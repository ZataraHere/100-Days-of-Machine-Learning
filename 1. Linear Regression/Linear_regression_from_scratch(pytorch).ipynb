```python
import torch
```

```python
class LinearRegressionTorch:

  def __init__(self):
    self.w = None
    self.b = None

  def initialize_params(self, X):
    self.w = torch.rand(1, X.shape[1], requires_grad=True)
    self.b = torch.zeros(1, requires_grad=True)

    return self.w,self.b

  def forward_propagation(self, X):

    return X @ self.w.T + self.b

  def fit_model(self, X, y, epochs=1000, Lr = 0.01):
    self.initialize_params(X)

    for epoch in range(epochs):
      y_pred = self.forward_propagation(X)
      loss = torch.mean((y_pred-y)**2)
      loss.backward()

      with torch.no_grad():
        self.w -= Lr * self.w.grad
        self.b -= Lr * self.b.grad
        self.w.grad.zero_()
        self.b.grad.zero_()

        if (epoch + 1) % 100 == 0:
                print(
                    f"Epoch {epoch + 1:4d} | "
                    f"Loss: {loss.item():.4f} | "
                    f"w: {self.w.detach()} | "
                    f"b: {self.b.item():.4f}"
                )

    return self.w, self.b

  def predict(self, X):
    return X @ self.w.T + self.b

  def evaluate(self, X, y):
    y_pred= self.predict(X)
    loss = torch.mean((y_pred-y)**2)

    return loss

```

```python
X = torch.tensor([
    [1.0, 5.0],
    [2.0, 6.0],
    [3.0, 5.0],
    [4.0, 7.0],
    [5.0, 6.0],
    [6.0, 8.0],
    [7.0, 7.0],
    [8.0, 8.0],
    [9.0, 7.0],
    [10.0, 9.0]
])

y = torch.tensor([
    [20.0],
    [27.0],
    [29.0],
    [39.0],
    [43.0],
    [51.0],
    [54.0],
    [62.0],
    [64.0],
    [73.0]
])
```

```python
print(X.shape)
print(y.shape)
```

```python
model = LinearRegressionTorch()

w, b = model.fit_model(
    X,
    y,
    epochs=1000,
    Lr=0.01
)
```

```python
loss = model.evaluate(X, y)

print("Final weights:", w)
print("Final bias:", b)
print("MSE:", loss.item())
```
