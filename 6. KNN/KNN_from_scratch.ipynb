```python
import numpy as np

# Features: [hours_studied, hours_slept]
X_train = np.array([
    [1, 5],
    [2, 6],
    [6, 7],
    [7, 8]
])

# Labels: 0 = Fail, 1 = Pass
y_train = np.array([0, 0, 1, 1])
```

```python

class KNN:
    def __init__(self, k=3):
        self.k = k

    def fit(self, X, y):
        # KNN stores the training data.
        self.X_train = np.asarray(X)
        self.y_train = np.asarray(y)

        if len(self.X_train) != len(self.y_train):
            raise ValueError("X and y must have the same number of samples.")

        if not 1 <= self.k <= len(self.X_train):
            raise ValueError("k must be between 1 and the number of samples.")

        return self

    def euclidean_distance(self, x1, x2):
        return np.sqrt(np.sum((x1 - x2) ** 2))

    def predict_one(self, x):
        # 1. Calculate distances to every training example.
        distances = np.array([
            self.euclidean_distance(x, x_train)
            for x_train in self.X_train
        ])

        # 2. Find indices of the k smallest distances.
        nearest_indices = np.argsort(distances)[:self.k]

        # 3. Get labels of the nearest neighbors.
        nearest_labels = self.y_train[nearest_indices]

        # 4. Return the most common label.
        labels, counts = np.unique(
            nearest_labels, return_counts=True
        )
        return labels[np.argmax(counts)]

    def predict(self, X):
        # Predict for multiple examples.
        return np.array([
            self.predict_one(x) for x in X
        ])


# Train
model = KNN(k=3)
model.fit(X_train, y_train)

# Predict
X_test = np.array([
    [5, 6],
    [1, 6],
    [7, 7]
])

predictions = model.predict(X_test)

print(predictions)
```
