```python
import numpy as np
```

```python
# Simple linearly separable dataset
X = np.array([
    [1, 2],
    [2, 3],
    [3, 3],
    [6, 5],
    [7, 8],
    [8, 6]
], dtype=float)

# Labels must be -1 or +1
y = np.array([-1, -1, -1, 1, 1, 1])
```

```python
class SVMfromScratch:

  def __init__(self,lr= 0.001, lambda_param = 0.01, n_iters = 2000):

    self.n_iters = n_iters
    self.lr = lr
    self.lambda_param = lambda_param
    self.w = None
    self.b = None

  def fit(self, X, y):

    n_samples, n_features = X.shape

    # initialize parameters

    self.w = np.zeros((n_features))
    self.b = 0.0

    # Gradient Descent

    for _ in range(self.n_iters):

       for idx, x_i in enumerate(X):
                condition = y[idx] * (
                    np.dot(x_i, self.w) + self.b
                ) >= 1

                if condition:
                    # Correct classification outside margin
                    dw = self.lambda_param * self.w
                    db = 0.0
                else:
                    # Misclassified or inside margin
                    dw = (
                        self.lambda_param * self.w
                        - y[idx] * x_i
                    )
                    db = -y[idx]


                self.w -= self.lr * dw
                self.b -= self.lr * db

  def predict(self, X):
        scores = np.dot(X, self.w) + self.b
        return np.where(scores >= 0, 1, -1)
```

```python
model = SVMfromScratch(
    lr = 0.001,
    lambda_param=0.01,
    n_iters=2000
)

model.fit(X, y)

print("Weights:", model.w)
print("Bias:", model.b)
```

```python
predictions = model.predict(X)

print("Predictions:", predictions)
print("Actual labels:", y)

accuracy = np.mean(predictions == y)
print("Accuracy:", accuracy)
```
